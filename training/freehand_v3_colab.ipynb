# Train freehand-v3

Trains Freehand's own model with [Soup](https://github.com/MakazhanAlpamys/Soup) on a free Colab GPU.

1. Check the top right says **T4**. If not: *Runtime → Change runtime type → T4 GPU*.
2. Click *Runtime → Run all*.
3. Wait about 1 to 2 hours. Keep this tab open.
4. At the end your browser downloads **freehand-v3-adapter.zip**. Leave it in your Downloads folder and tell Claude it is done.

```python
# 1. Check the GPU
!nvidia-smi --query-gpu=name,memory.total --format=csv
```

```python
# 2. Install Soup (keeps Colab's own PyTorch)
!pip -q install "soup-cli[train]"
!soup --version
```

```python
# 3. Get the training data and settings from the Freehand repo
!wget -q -O freehand.jsonl https://raw.githubusercontent.com/Jayswansocool48/freehand/main/training/freehand.jsonl
!wget -q -O colab.yaml https://raw.githubusercontent.com/Jayswansocool48/freehand/main/training/colab.yaml
!wc -l freehand.jsonl
```

```python
# 4. Train (about 1 to 2 hours)
!soup train --config colab.yaml --yes
```

```python
# 5. Download the result
!cd output && zip -q -r ../freehand-v3-adapter.zip adapter_config.json adapter_model.safetensors
from google.colab import files
files.download('freehand-v3-adapter.zip')
```
